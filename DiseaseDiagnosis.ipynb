```python
import pandas as pd
data_set = pd.read_csv('DiseaseAndSymptoms.csv')
```

```python
symptoms_dataframe = data_set.drop('Disease',axis=1)
symptoms_dataframe = symptoms_dataframe.fillna('None')

first_list = [ ]
for row_index in symptoms_dataframe.index:
  for column_name in symptoms_dataframe.columns:
    cell_value = symptoms_dataframe.loc[row_index,column_name]
    cell_value = cell_value.strip()
    if(cell_value != 'None'):
      first_list.append((row_index,column_name,cell_value))

```

```python
stacked_dataframe = pd.DataFrame(first_list)
stacked_dataframe[3] = 1
display(stacked_dataframe)


#This is encoding with 1
indra_chowk = pd.pivot_table(data=stacked_dataframe,index=0,columns=2,values=3)
#This is encoding with 0
indra_chowk = indra_chowk.fillna('0')
indra_chowk = indra_chowk.astype(int)
display(indra_chowk)
```

```python
disease = data_set['Disease']
from sklearn.preprocessing import LabelEncoder
disease_encoder = LabelEncoder()
encoding_disease = disease_encoder.fit_transform(disease)
encoded_disease = pd.Series(encoding_disease)

import warnings
warnings.filterwarnings('ignore')
indra_chowk.insert(loc=0,column='Disease',value=encoded_disease)
```
