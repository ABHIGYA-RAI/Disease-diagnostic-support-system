```python
import pandas as pd
data_set = pd.read_csv('DiseaseAndSymptoms.csv')
```

```python
symptoms_dataframe = data_set.drop('Disease',axis=1)
symptoms_dataframe = symptoms_dataframe.fillna('None')

first_list = [ ]
for row_index in symptoms_dataframe.index:
  for column_name in symptoms_dataframe.columns:
    cell_value = symptoms_dataframe.loc[row_index,column_name]
    cell_value = cell_value.strip()
    if(cell_value != 'None'):
      first_list.append((row_index,column_name,cell_value))

```

```python
stacked_dataframe = pd.DataFrame(first_list)
stacked_dataframe[3] = 1
display(stacked_dataframe)


#This is encoding with 1
indra_chowk = pd.pivot_table(data=stacked_dataframe,index=0,columns=2,values=3)
#This is encoding with 0
indra_chowk = indra_chowk.fillna('0')
indra_chowk = indra_chowk.astype(int)
display(indra_chowk)
```

```python
disease = data_set['Disease']
from sklearn.preprocessing import LabelEncoder
disease_encoder = LabelEncoder()
encoding_disease = disease_encoder.fit_transform(disease)
encoded_disease = pd.Series(encoding_disease)

import warnings
warnings.filterwarnings('ignore')
indra_chowk.insert(loc=0,column='Disease',value=encoded_disease)
```

```python
no_underscore_list = [ ]
for e in kaggle_symptoms:
    no_underscore = e.replace('_',' ')
    no_underscore_list.append(no_underscore)
```

```python
kaggle_symptoms_to_match = [ ] 
for k in no_underscore_list:
    if k not in exact_matches:
        kaggle_symptoms_to_match.append(k)
print(len(kaggle_symptoms_to_match))
```

```python
prompt = f"""Match each symptoms in LIST A with the closest equivalent in LIST B. 
Don't try to forcefully match the symptoms in LIST A with LIST B.
Only match if they are similar or convey the same meaning.
If no equivalent exists, just use null.
During matching, use the exact words that are in LIST B.
Return the answer in JSON format: {{"<LIST A symptom>" : "<LIST B symptom or null>"}}

LIST A : {kaggle_symptoms_to_match}
LIST B : {mendeley_symptoms_list}
"""

try:
    response = client.models.generate_content(
        model = "gemini-3.5-flash-lite",
        contents = prompt,
        config = {"response_mime_type" : "application/json"}
    )
    gemini_map = json.loads(response.text)
    print(len(gemini_map))
except Exception as e:
    print(f"Sorry, there's an error: {e}")
```

```python
training_disease = indra_chowk['Disease']
training_symptoms = indra_chowk.drop('Disease',axis=1)
```

```python
import sklearn
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import BernoulliNB
from sklearn.naive_bayes import MultinomialNB
from sklearn.naive_bayes import GaussianNB
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression
```

```python
classifier_dictionary = {
    "K-Nearest Neighbor" : KNeighborsClassifier(),
    "Support Vector Machine" : SVC(),
    "Random Forest" : RandomForestClassifier(),
    "Bernoulli Naive Bayes" : BernoulliNB(),
    "Multinomial Naive Bayes" : MultinomialNB(),
    "Gaussian Naive Bayes" : GaussianNB(),
    "Cat Boost" : CatBoostClassifier(),
    "Light Boost" : LGBMClassifier(),
    "XG Boost" : XGBClassifier(),
    "Logistic Regression" : LogisticRegression()
    
}
```

```python
for name,machine in classifier_dictionary.items():
    machine.fit(training_symptoms,training_disease)
```

```python
for name,machine in classifier_dictionary.items():
    diagnosis = machine.predict(testing_dataframe.values)
```

```python
from sklearn.metrics import accuracy_score
for name,machine in classifier_dictionary.items():
    accuracy = accuracy_score(testing_disease,diagnosis)
    print(f"{name} : {accuracy}")
```
