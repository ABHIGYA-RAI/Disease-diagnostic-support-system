```python
import pandas as pd
data_set = pd.read_csv('/content/DiseaseAndSymptoms.csv')
```

```python
symbi_predict = pd.read_csv('/content/symbipredict_2022.csv')
print(symbi_predict)
```

```python
symptoms_dataframe = data_set.drop('Disease',axis=1)
symptoms_dataframe = symptoms_dataframe.fillna('None')
```

```python
first_list = [ ]
for row_index in symptoms_dataframe.index:
  for column_name in symptoms_dataframe.columns:
    cell_value = symptoms_dataframe.loc[row_index,column_name]
    cell_value = cell_value.strip()
    if(cell_value != 'None'):
      first_list.append((row_index,column_name,cell_value))
print(first_list)
```

```python
stacked_dataframe = pd.DataFrame(first_list)
stacked_dataframe[3] = 1
display(stacked_dataframe)
```

```python
#This is encoding with 1
indra_chowk = pd.pivot_table(data=stacked_dataframe,index=0,columns=2,values=3)
#This is encoding with 0
indra_chowk = indra_chowk.fillna('0')
indra_chowk = indra_chowk.astype(int)
display(indra_chowk)
```

```python
disease = data_set['Disease']
from sklearn.preprocessing import LabelEncoder
disease_encoder = LabelEncoder()
encoding_disease = disease_encoder.fit_transform(disease)
encoded_disease = pd.Series(encoding_disease)
display(encoded_disease)
```

```python
import warnings
warnings.filterwarnings('ignore')
indra_chowk.insert(loc=0,column='Disease',value=encoded_disease)
display(indra_chowk)
```

```python
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.naive_bayes import MultinomialNB
from sklearn.naive_bayes import BernoulliNB
from catboost import CatBoostClassifier
from sklearn.ensemble import RandomForestClassifier
from lightgbm import LGBMClassifier

classifier_dictionary = {
    #"Support Vector Machine" : SVC(),
    "K-Nearest Neighbor" : KNeighborsClassifier(),
    "Logistic Regression" : LogisticRegression(),
    "Gaussian Naive Bayes" : GaussianNB(),
    "Multinomial Naive Bayes" : MultinomialNB(),
    "Bernoulli Naive Bayes" : BernoulliNB(),
    "Random Forest" : RandomForestClassifier(),
    "Light Boost" : LGBMClassifier()
}
```

```python
# Categorizing the symptoms and diseases to train the classifiers
symptom_feature = indra_chowk.drop('Disease',axis=1)
disease_target = indra_chowk['Disease']
```

```python
for name_of_classifier,classifier in classifier_dictionary.items():
  classifier.fit(symptom_feature,disease_target)
```

```python
matched_list = [ ]
unmatched_list = [ ]
training_list = list(indra_chowk.columns)
training_list.remove('Disease')
testing_list = list(symbi_predict.columns)
testing_list.remove('prognosis')
for e in testing_list:
  if(e in training_list):
    matched_list.append(e)
  elif(e not in training_list):
    unmatched_list.append(e)
'''print(matched_list)
print(unmatched_list)'''
print(training_list)
```

```python
#Arranging the order of the columns
wow = symbi_predict[training_list]
display(wow)
```

```python
disease_rename_map = {
    "Fungal Infection" : "Fungal infection",
    "Chronic Cholestasis" : "Chronic cholestasis",
    "Cervical Spondylosis" : "Cervical spondylosis",
    "Hepatitis A" : "hepatitis A",
    "Alcoholic Hepatitis" : "Alcoholic hepatitis",
    "Heart Attack" : "Heart attack",
    "Varicose Veins" : "Varicose veins",
    "Urinary Tract Infection" : "Urinary tract infection",
    "Chickenpox" : "Chicken pox",
    "Peptic Ulcer Disease" : "Peptic ulcer diseae",
    "Dimorphic Hemmorhoids (piles)" : "Dimorphic hemmorhoids(piles)",
    "Osteoarthritis" : "Osteoarthristis",
    "Vertigo" : "(vertigo) Paroymsal  Positional Vertigo"
}
```

```python
just_checking = symbi_predict['prognosis']
just_checking = just_checking.replace(disease_rename_map)
display(just_checking)
```

```python
encoding_symbi = disease_encoder.transform(just_checking)
encoded_symbi_disease = pd.Series(encoding_symbi)
display(encoded_symbi_disease)
```

```python
import numpy as np
from sklearn.metrics import accuracy_score
def corrupt(df, flip_frac=0.15, seed=42):
    rng = np.random.default_rng(seed)
    mask = rng.random(df.shape) < flip_frac
    return df ^ mask.astype(int)

X_test_noisy = corrupt(wow)
name_list = [ ]
noisy_list = [ ]
clean_list = [ ]
difference_list = [ ]

for name, model in classifier_dictionary.items():
    name_list.append(name)
    acc_clean = accuracy_score(encoded_symbi_disease, model.predict(wow))
    clean_list.append(acc_clean)
    acc_noisy = accuracy_score(encoded_symbi_disease, model.predict(X_test_noisy))
    noisy_list.append(acc_noisy)
    accuracy_difference = (acc_clean-acc_noisy)
    difference_list.append(accuracy_difference)
comparison_dictionary = {
        "Classifier" : name_list,
        "Noisy symptoms accuracy" : noisy_list,
        "Clean symptoms accuracy" : clean_list,
        "Accuracy difference" : difference_list
    }
ranjana_lipi = pd.DataFrame(comparison_dictionary)
ranjana_lipi = ranjana_lipi.sort_values(by="Accuracy difference",ascending=True)
display(ranjana_lipi)


```

```python
from sklearn.metrics import accuracy_score
def corrupt(df,seed=10,flip_frac=0.15):
  rng = np.random.default_rng(seed)
  mask = rng.random(df.shape) < flip_frac
  return df ^ mask.astype(int)



corrupted = corrupt(wow)
name_list = [ ]
clean_accuracy_list = [ ]
corrupted_accuracy_list = [ ]
accuracy_difference_list = [ ]
for classifier_name,classifier in classifier_dictionary.items():
  name_list.append(classifier_name)
  clean_accuracy = accuracy_score(encoded_symbi_disease,classifier.predict(wow))
  clean_accuracy_list.append(clean_accuracy)
  corrupted_accuracy = accuracy_score(encoded_symbi_disease,classifier.predict(corrupted))
  corrupted_accuracy_list.append(corrupted_accuracy)
  accuracy_difference = (clean_accuracy-corrupted_accuracy)
  accuracy_difference_list.append(accuracy_difference)
cutie_antu = {
    "Name of classifier" : name_list,
    "Clean accuracy" : clean_accuracy_list,
    "Corrupted accuracy" : corrupted_accuracy_list,
    "Difference in accuracy" : accuracy_difference_list
}

comparison_dataframe = pd.DataFrame(cutie_antu)
comparison_dataframe = comparison_dataframe.sort_values(by="Difference in accuracy",ascending=True)
print(comparison_dataframe)

```

```python
import numpy as np
import pandas as pd

def corrupt(df, flip_frac=0.15, seed=42):
    rng = np.random.default_rng(seed)
    mask = rng.random(df.shape) < flip_frac
    return df ^ mask.astype(int)

results = []
for frac in [0.02, 0.05, 0.10, 0.15]:
    X_test_noisy = corrupt(wow, flip_frac=frac)
    for name, model in classifier_dictionary.items():
        acc = accuracy_score(encoded_symbi_disease, model.predict(X_test_noisy))
        results.append({"flip_frac": frac, "model": name, "accuracy": round(acc, 4)})

results_df = pd.DataFrame(results)
pivot = results_df.pivot(index="model", columns="flip_frac", values="accuracy")
display(pivot)
```

```python
import numpy as np
beabadobee = [ ]
def corrupt(df,flip_frac=0.15,seed=42):
  rng = np.random.default_rng(seed)
  mask = rng.random(df.shape) < flip_frac
  return df ^ mask.astype(int)

for fraction in [0.02,0.05,0.10,0.15]:
  abcd = corrupt(wow,flip_frac=fraction)
  for name,machine in classifier_dictionary.items():
    heram_accuracy = accuracy_score(encoded_symbi_disease,machine.predict(abcd))
    beabadobee.append({"Flipped Fraction" : fraction, "Machine" : name, "Actual Accuracy" : round(heram_accuracy,4)})

df_two = pd.DataFrame(beabadobee)
df_three = df_two.pivot(index = "Machine", columns= "Flipped Fraction", values = "Actual Accuracy")
display(df_three)

```
