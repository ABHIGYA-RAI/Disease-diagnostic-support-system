```python
from catboost import CatBoostClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
#AdaBoost
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import AdaBoostClassifier
```

```python
classifier_dictionary = {
    "Support Vector Machine" : SVC(),
    "Logistic Regression" : LogisticRegression(),
    "Random Forest" : RandomForestClassifier(),
    "K Nearest Neighbor" : KNeighborsClassifier(),
    "Gaussian Naive Bayes" : GaussianNB(),
    "Multinomial Naive Bayes" : MultinomialNB(),
    "Ada Useless Boost" : AdaBoostClassifier(),
    "Light GBM" : LGBMClassifier(),
    "XG Boost" : XGBClassifier(),
    "Cat Boost" : CatBoostClassifier()
}
```

```python
for classifier_name,classifier in classifier_dictionary.items():
  classifier.fit(training_symptom,training_disease)
```

```python
import warnings
warnings.filterwarnings('ignore')
accuracy_list = [ ]
precision_list = [ ]
recall_list = [ ]
f1_list = [ ]
classifier_name_list = [ ]
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import f1_score
from sklearn.metrics import recall_score
for classifier_name,classifier in classifier_dictionary.items():
  diagnosis = classifier.predict(testing_symptom)
  classifier_name_list.append(classifier_name)
  accuracy = accuracy_score(testing_disease,diagnosis)
  accuracy_list.append(accuracy)
  precision = precision_score(testing_disease,diagnosis,average='macro')
  precision_list.append(precision)
  f1 = f1_score(testing_disease,diagnosis,average='macro')
  f1_list.append(f1)
  recall = recall_score(testing_disease,diagnosis,average='macro')
  recall_list.append(recall)
evaluation_dictionary = {
    "Classifier" : classifier_name_list,
    "Accuracy Score" : accuracy_list,
    "Precision Score" : precision_list,
    "F1 Score" : f1_list,
    "Recall Score" : recall_list
}
evaluation_dataframe = pd.DataFrame(evaluation_dictionary)
display(evaluation_dataframe)
```
